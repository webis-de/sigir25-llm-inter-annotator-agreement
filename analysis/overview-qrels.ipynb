```python
from trectools import TrecQrel

def report_for_single_qrel_file(qrel):
    ret = []
    for rel in [0, 1, 2, 3]:
        ret += [len(qrel[qrel['rel'] == rel])/len(qrel)]
    return ret

def report_qrels(method):
    qrels = [TrecQrel(f'../data/msmarco-passage-trec-dl-20{i}-judged/qrels/{method}.qrels.txt') for i in ['19', '20']]

    ret = []
    for qrel in qrels:
        ret += report_for_single_qrel_file(qrel.qrels_data)

    ret = ["{:.3f}".format(i) for i in ret]
    return ' & '.join([i.replace('0.', '.') for i in ret])
```
